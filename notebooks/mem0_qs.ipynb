## 定义配置


```python
from mem0 import Memory

config = {
    "vector_store": {
        "provider": "qdrant",
        "config": {"host": "localhost", 
                   "port": 6333,
                   "embedding_model_dims": 512
                   },
    },
    "llm": {
        "provider": "ollama",
        "config": {"model": "minimax-m3:cloud", "temperature": 0.1},
    },
    "embedder": {
        "provider": "huggingface",
        "config": {"model": "BAAI/bge-small-zh-v1.5"},
    },
    "reranker": {
        "provider": "huggingface",
        "config": {"model": "BAAI/bge-reranker-base"},
    },
}

memory = Memory.from_config(config)
```

```python
messages = [
    {"role": "user", "content": "I'm planning to watch a movie tonight. Any recommendations?"},
    {"role": "assistant", "content": "How about thriller movies? They can be quite engaging."},
    {"role": "user", "content": "I'm not a big fan of thriller movies but I love sci-fi movies."},
    {"role": "assistant", "content": "Got it! I'll avoid thriller recommendations and suggest sci-fi movies in the future."}
]
memory.add(messages, user_id="alice", metadata={"category": "movies"})
```

```python
memories = memory.get_all(filters={"user_id": "alice"})
print(memories)
```
